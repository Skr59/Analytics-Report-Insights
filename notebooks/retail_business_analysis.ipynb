## Average Order Value (AOV)

Average Order Value is calculated by dividing the total transaction
sales value by the number of unique invoices.

AOV = Total Transaction Sales / Number of Unique Invoices

This is a preliminary measure. Cancellation and return treatment must
be finalized before reporting the final AOV.

```python
customer_orders = (
    customer_df.groupby("Customer ID")["Invoice"]
    .nunique()
)

repeat_customers = (customer_orders > 1).sum()
total_customers = customer_orders.count()

repeat_purchase_rate = (
    repeat_customers / total_customers
) * 100

print("Total customers:", total_customers)
print("Repeat customers:", repeat_customers)
print("Repeat purchase rate:", round(repeat_purchase_rate, 2), "%")
```

## Repeat Purchase Rate

Repeat purchase rate is calculated as the percentage of identified
customers who have placed more than one unique invoice.

Repeat Purchase Rate =
(Repeat Customers / Total Identified Customers) × 100

This is a transaction-based measure and does not establish why
customers returned.

```python
total_sales = analysis_df["Sales"].sum()

product_summary = (
    analysis_df.groupby("Description")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

product_summary["Sales Contribution (%)"] = (
    product_summary["Sales"] / total_sales
) * 100

print(product_summary.head(10))
```

This is a transaction-based retention measure. It does not explain

```python
final_customer_sales = (
    final_customer_df.groupby("Customer ID")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 Customers by Positive Sales:")
display(final_customer_sales.head(10).reset_index())
```

## Final Customer Sales Analysis

Customer-level positive transaction sales are aggregated and ranked
in descending order.

Only transactions with valid Customer IDs are included.

These values represent transaction sales, not customer profitability
or lifetime value.

```python
top_10_customers = (
    final_customer_sales.head(10)
    .sort_values(ascending=True)
)

plt.figure(figsize=(12, 7))

plt.barh(
    top_10_customers.index.astype(str),
    top_10_customers.values
)

plt.title("Top 10 Customers by Positive Transaction Sales")
plt.xlabel("Sales Value")
plt.ylabel("Customer ID")
plt.tight_layout()
plt.show()
```

## Top 10 Customer Visualization

positive transaction sales values.

This visualization supports customer segmentation and retention

```python
country_summary = (
    final_sales_df.groupby("Country")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

country_summary["Sales Contribution (%)"] = (
    country_summary["Sales"] /
    final_sales_df["Sales"].sum()
) * 100

display(country_summary.head(10))
```

## Country-Wise Sales Contribution

Country-level positive transaction sales are calculated as a percentage
of total positive transaction sales.

Sales Contribution (%) =
(Country Sales / Total Positive Sales) × 100

The results help assess geographic sales concentration. They do not
represent country-level profitability.

```python
top_10_countries = (
    country_summary.head(10)
    .sort_values(by="Sales Contribution (%)", ascending=True)
)

plt.figure(figsize=(12, 7))

plt.barh(
    top_10_countries["Country"],
    top_10_countries["Sales Contribution (%)"]
)

plt.title("Top 10 Countries by Sales Contribution")
plt.xlabel("Sales Contribution (%)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```
